## 3. Automated HTML EDA Report (With `ydata-profiling`)

Having browsed through multiple data related articles to streamline my workflow for future projects, I chanced upon a tool that is extremely useful for EDA. `ydata-profiling` provides an efficient way to understand data, by generating an interactive reports, which highlight data types, with little coding effort. From the report, I can easily identify flaws in datasets like missing values, duplicates, class imbalance etc. I can also visualise the skewness of each individual column. This greatly expedites tedious EDA processes, and allow data scientists to reduce time-to-production for ML models. However, it is still needed for me to write seaborn / matplotlib code in order to gain more granular insights into our dataset.


```python
# - Inspect Dataframe - 
df1 = spark.read.csv("data/raw/creditcard.csv", header=True, inferSchema=True)
print(f"Dataframe shape: {df1.count()}, {len(df1.columns)}")
display(df1.limit(5).toPandas())
```

```python
# - Pandas Profiling Report -
type_schema = {"Class": "categorical"}

profile = ProfileReport(df1,title="EDA Report (Fraud Detection Dataset)", 
                        type_schema=type_schema,
                        correlations={
                              "pearson": {
                                    "calculate": True, 
                                    "warn_high_correlations": True,
                                    "threshold": 0.65}},
                        
                        explorative=True)

profile.to_file("eda_report.html")
```
